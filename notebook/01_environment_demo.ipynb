# Environment demo

This notebook demonstrates the `MazeEnv` environment: reset, stepping, and a small random episode.
Run the cells below to interactively explore the environment.

```python
import sys
from pathlib import Path

# Add project root to path so imports work
project_root = Path("..").resolve()
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))
```

```python
# Imports and environment creation
from envs import MazeEnv

# Create an easy maze and reset
env = MazeEnv(difficulty="easy")
obs, info = env.reset()
print('start state:', obs, 'start pos:', info['position'])
```

```python
# Step a few random actions and show info
for action in range(4):
    obs, reward, terminated, truncated, info = env.step(action)
    print(f'action={action}: pos={info["position"]} reward={reward:.2f} terminated={terminated}')
```

### Visualizing the maze
The environment includes a matplotlib-based renderer. To display the maze in a notebook, run the next cell.

```python
# Render the current maze (this will open a matplotlib figure inline if running in a Jupyter environment)
env.render_mode = "human"
env.render()
```
